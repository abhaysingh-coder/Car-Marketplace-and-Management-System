```python
data = pd.read_csv('../Data.csv')
data.drop(['Engine Fuel Type', 'Number of Doors'], axis=1, inplace=True)
data.columns = [col.replace(" ", "_") for col in data.columns]
```

```python
data.shape
```

```python
col =['Car_ID']
for i in data.columns:
    col.append(i)
```

```python
Buy_Data = pd.DataFrame(columns=col)
Buy_Request = pd.DataFrame(columns= col)
Buy_History = pd.DataFrame(columns= col)
```

```python
for i in range(len(data)):
    Buy_Data = pd.concat([Buy_Data, data.iloc[[i]]], ignore_index=True)
    Buy_Data['Car_ID'] = range(1, len(Buy_Data) + 1)
```

```python
Buy_Data.shape
```

```python
for i in range(10):
    Buy_Request = pd.concat([Buy_Request, Buy_Data.iloc[[0]]], ignore_index=True)
    Buy_Data.drop(Buy_Data.index[0], inplace=True)
    Buy_Data['Buy_Request'] = range(1, len(Buy_Data) + 1)
```

```python
Buy_Data.shape
```

```python
Buy_Request.shape
```

```python
for i in range(10):
    Buy_History = pd.concat([Buy_History, Buy_Data.iloc[[0]]], ignore_index=True)
    Buy_Data.drop(Buy_Data.index[0], inplace=True)    
```

```python
Buy_Data.shape
```

```python
Buy_History.shape
```

```python
Buy_Data.to_csv('../Buy Data.csv', index=False)
Buy_Request.to_csv('../Buy Request.csv', index=False)
Buy_History.to_csv('../Buy History.csv', index=False)
```
