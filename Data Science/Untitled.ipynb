```python
import os
html_folder = "../mainapp/templates"  # Change to your HTML folder
output_file = "all_html_code.txt"

with open(output_file, "w", encoding="utf-8") as outfile:

    for root, dirs, files in os.walk(html_folder):

        for file in files:

            if file.endswith(".html"):

                file_path = os.path.join(root, file)

                outfile.write("=" * 100 + "\n")
                outfile.write(f"FILE: {file_path}\n")
                outfile.write("=" * 100 + "\n\n")

                try:
                    with open(file_path, "r", encoding="utf-8") as infile:
                        outfile.write(infile.read())

                except Exception as e:
                    outfile.write(f"\nERROR READING FILE: {e}\n")

                outfile.write("\n\n\n")

print(f"All HTML files saved to '{output_file}'")
```

```python
import os
output_file = "all_html_code.txt"
folder = os.path.join('..', 'mainapp', 'templates')
with open(output_file, "w", encoding="utf-8") as outfile:
    for img in os.listdir(folder):
        try:
            html_path = os.path.join(folder, img)
            with open(html_path, "r", encoding="utf-8") as file:
                content = file.read()
                outfile.write('html_page ' + img + '\n')                
                outfile.write(content)
                outfile.write('\n\n\n\n')
        except Exception as e:
            print(e)
```
